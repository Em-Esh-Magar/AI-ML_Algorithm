```python
import numpy as pd
import pandas as pd
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_blobs
from sklearn.metrics import silhouette_score
from sklearn.datasets import make_moons
```

```python
X, Y = make_moons(n_samples=750, noise=0.10, random_state=42)
plt.scatter(X[:,0], X[:,1])
```

```python
dbscan = DBSCAN(eps=0.2, min_samples=5)
```

```python
y_predicted = dbscan.fit_predict(X)
```

```python
y_predicted
```

```python
sil_score = silhouette_score(X,y_predicted)
```

```python
sil_score
```

```python
n_clusters = len(set(y_predicted)) - (1 if -1 in y_predicted else 0)
n_noise = list(y_predicted).count(-1)

print(f"Clusters found : {n_clusters}")
print(f"Noise found : {n_noise}")
```
